```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns


# 1. Load the file
df = pd.read_excel(r"D:\PowerBI_Blank_Table.xlsx.csv.xlsx")

# 2. Check rows & columns
print("Rows and Columns:", df.shape)

# 3. View first 5 rows
df.head()
```

```python
print(df.columns.tolist())
```

```python
# Q15: Tenure in years
df["customer_tenure_years"] = (df["tenure_months"] / 12).round(2)

# Convert dates into datetime format
df["signup_date"] = pd.to_datetime(df["signup_date"], errors="coerce")
df["last_active_date"] = pd.to_datetime(df["last_active_date"], errors="coerce")

# Q16: Total active days
df["days_active"] = (
    df["last_active_date"] - df["signup_date"]
).dt.days

# Verify results
df.head()
```

```python
# Q15: Tenure in years
df["customer_tenure_years"] = (df["tenure_months"] / 12).round(2)

# Q16: Total active days between signup and last activity
df["days_active"] = (df["last_active_date"] - df["signup_date"]).dt.days

# Q17 & Q18: Signup month and year
df["signup_month"] = df["signup_date"].dt.month
df["signup_year"] = df["signup_date"].dt.year

# Check output
df[["tenure_months", "customer_tenure_years", "days_active", "signup_month", "signup_year"]].head()
```

```python
# Q19: Average revenue generated per month
df["revenue_per_month"] = np.where(
    df["tenure_months"] > 0,
    (df["total_revenue"] / df["tenure_months"]).round(2),
    df["monthly_fee"]
)

# Q20: Flag top 25% revenue customers (1 = High Value, 0 = Regular)
df["high_value_customer"] = (df["total_revenue"] > df["total_revenue"].quantile(0.75)).astype(int)

# Q21: Categorize total revenue into Low / Medium / High
df["revenue_category"] = pd.cut(
    df["total_revenue"],
    bins=[-np.inf, 7000, 20000, np.inf],
    labels=["Low", "Medium", "High"]
)

# Check output
df[["total_revenue", "revenue_per_month", "high_value_customer", "revenue_category"]].head()
```

```python
# Q22: Age brackets
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 30, 45, 60, 100],
    labels=["18-30", "31-45", "46-60", "60+"]
)

# Q23: Session duration categories (minutes)
df["session_category"] = pd.cut(
    df["avg_session_time"],
    bins=[-np.inf, 40, 80, np.inf],
    labels=["Low", "Medium", "High"]
)

# Q24: Discount tiers
df["discount_category"] = pd.cut(
    df["discount_percent"],
    bins=[-1, 0, 10, 100],
    labels=["None", "Low", "High"]
)

# Q25: Support ticket load
df["support_intensity"] = pd.cut(
    df["support_tickets"],
    bins=[-1, 1, 3, 100],
    labels=["Low", "Medium", "High"]
)

# Check output
df[["age_group", "session_category", "discount_category", "support_intensity"]].head()
```

```python
# Q26: Binary churn target (1 = Yes, 0 = No)
df["churn_flag"] = (df["churned"] == "Yes").astype(int)

# Q27: Ordinal encoding for subscription plan
df["plan_rank"] = df["plan_type"].map({"Basic": 1, "Standard": 2, "Premium": 3})

# Q28: Engagement relative to support friction
df["engagement_score"] = (df["avg_session_time"] / (df["support_tickets"] + 1)).round(2)

# Check output
df[["churned", "churn_flag", "plan_type", "plan_rank", "engagement_score"]].head()
```

```python
from pathlib import Path

output_dir = Path("../output")
output_dir.mkdir(parents=True, exist_ok=True)

# Save cleaned + feature-engineered dataset
df.to_csv(output_dir / "cleaned_data.csv", index=False)

print("Final shape:", df.shape)
print("Total missing values:", df.isna().sum().sum())
print("Saved to:", (output_dir / "cleaned_data.csv").resolve())
```
