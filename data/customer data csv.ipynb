```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns


# 1. Load the file
df = pd.read_excel(r"D:\PowerBI_Blank_Table.xlsx.csv.xlsx")

# 2. Check rows & columns
print("Rows and Columns:", df.shape)

# 3. View first 5 rows
df.head()

```

```python
# 1. Print all Q1-Q4 checks cleanly in one cell
print("Q1 Shape:", df.shape)
print("\nQ2 Data Types:\n", df.dtypes)
print("\nQ3 Missing Count:\n", df.isna().sum())
print("\nQ4 Missing %:\n", (df.isna().mean() * 100).round(2))

# 2. Fix Q2 issue: Convert text dates ('object') to real datetime format
df["signup_date"] = pd.to_datetime(df["signup_date"], format="mixed")
df["last_active_date"] = pd.to_datetime(df["last_active_date"], format="mixed")

# 3. Fix Q3 & Q4 issue: Fill the ~4% missing values
df["age"] = df["age"].fillna(df["age"].median())
df["avg_session_time"] = df["avg_session_time"].fillna(df["avg_session_time"].median())
df["discount_percent"] = df["discount_percent"].fillna(0)
df["city"] = df["city"].fillna("Unknown")

# Verify 0 missing values remain
print("\nRemaining Missing Values:", df.isna().sum().sum())
```

```python
import numpy as np

# Q9: Count exact duplicate rows
print("Duplicate rows found:", df.duplicated().sum())

# Q10: Remove duplicates and confirm new row count
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
# Q11: Convert text dates into real datetime objects
df["signup_date"] = pd.to_datetime(df["signup_date"], format="mixed", errors="coerce")
df["last_active_date"] = pd.to_datetime(df["last_active_date"], format="mixed", errors="coerce")

# Verify conversion and check if any unparseable dates (NaT) appeared
print(df[["signup_date", "last_active_date"]].dtypes)
print("Unparseable dates (NaT):", df[["signup_date", "last_active_date"]].isna().sum().to_dict())
```

```python
# Q12: Count rows where last_active_date happens BEFORE signup_date
illogical_dates = (df["last_active_date"] < df["signup_date"]).sum()
print("Illogical date rows found:", illogical_dates)

# Q13: Keep only valid records where last_active_date >= signup_date
df = df[df["last_active_date"] >= df["signup_date"]]
print("Shape after removing illogical dates:", df.shape)
```

```python
# Q14: Check for unrealistic ages (< 18 or > 80) and set them to NaN
invalid_ages = ((df["age"] < 18) | (df["age"] > 80)).sum()
print("Invalid ages found:", invalid_ages)

df.loc[(df["age"] < 18) | (df["age"] > 80), "age"] = np.nan
```

```python
# Q5: Fill missing age with median age
df["age"] = df["age"].fillna(df["age"].median())

# Q6: Fill missing city with "Unknown"
df["city"] = df["city"].fillna("Unknown")

# Q7: Fill missing avg_session_time with median session time
df["avg_session_time"] = df["avg_session_time"].fillna(df["avg_session_time"].median())

# Q8: Fill missing discount_percent with 0
df["discount_percent"] = df["discount_percent"].fillna(0)
```

```python
# Confirm 0 missing values remain across all 15 columns
print("Missing values after cleaning:\n", df.isna().sum())
print("\nFinal clean dataset shape:", df.shape)

# Optional: Save the cleaned dataset to data/processed/ without touching your raw file
df.to_csv("PowerBI_Clean_Table.csv", index=False)
```
